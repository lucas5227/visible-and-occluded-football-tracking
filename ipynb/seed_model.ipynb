```python
import torch
from ultralytics import YOLO

# Device 선택
if torch.cuda.is_available():
    device = 0
    device = 1
elif torch.backends.mps.is_available():
    device = "mps"
else:
    device = "cpu"

print(f"Using device: {device}")

# 모델 로드
model = YOLO("yolo26m.pt")

results = model.train(
    data="../data/AFMOT_seed/data.yaml",

    # 학습
    epochs=100,
    patience=20,

    # 작은 공 대응
    imgsz=1280,

    # 배치
    batch=8 if device == "mps" else 16,

    # 장치
    device=device,

    # 증강
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    degrees=5,
    translate=0.1,
    scale=0.3,

    fliplr=0.5,
    mosaic=1.0,
    mixup=0.0,

    # 캐시
    cache=True,

    # 저장
    project="runs_seed",
    name="yolo26m_afmot_seed",

    workers=8
)

print("Saved to:", results.save_dir)
```
